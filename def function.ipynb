```python
def table(num):
    for i in range(1,11):
        print(num*i)
def repeat(num):
    for i in range(num):
        print(f"repeat this number {num} times u give in option")

option=int(input("option: "))
num=int(input("enter a number: "))

if option==1:
    table(num)
elif option==2:
    repeat(num)
else:
    print("please check again error showing your correctly give a option 1 or 2!")
```
