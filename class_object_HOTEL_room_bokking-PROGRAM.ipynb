```python
class Normal_rooms:
    def calculates(self, no_rooms, days):
        if days == 1:
            rent = 300
        elif 1 < days <= 5:
            rent = 250
        elif days > 5:
            rent = 200
        else:
            rent = 0
        return rent * no_rooms  # Return the rent instead of printing it

class AC_room(Normal_rooms):
    def AC_calculates(self, no_rooms, days):
        if days == 1:
            rent = 450
        elif 1 < days <= 5:
            rent = 300
        elif days > 5:
            rent = 250
        else:
            rent = 0
        return rent * no_rooms  # Return the rent instead of printing it

class Suite_room(AC_room):
    def Suite_calculates(self, no_rooms, days):
        if days == 1:
            rent = 550
        elif 1 < days <= 5:
            rent = 500
        elif days > 5:
            rent = 450
        else:
            rent = 0
        return rent * no_rooms  # Return the rent instead of printing it

class Hotel(Suite_room):
    def __init__(self, name, address, mobile):
        self.name = name
        self.address = address
        self.mobile = mobile

    def display(self):
        print(f"Name: {self.name}")
        print(f"Address: {self.address}")
        print(f"Mobile Number: {self.mobile}")

# Input and Usage
name = input("Enter your name: ")
address = input("Enter your address: ")
mobile = input("Enter your mobile number: ")
no_rooms = int(input("Enter the number of rooms: "))
days = int(input("Enter the number of days: "))

hotel = Hotel(name, address, mobile)
hotel.display()

room_type = input("Enter the room type (Normal/AC/Suite): ").lower()

if room_type == "normal":
    total_rent = hotel.calculates(no_rooms, days)
    print(f"The total rent for Normal room(s) is: {total_rent}")
elif room_type == "ac":
    total_rent = hotel.AC_calculates(no_rooms, days)
    print(f"The total rent for AC room(s) is: {total_rent}")
elif room_type == "suite":
    total_rent = hotel.Suite_calculates(no_rooms, days)
    print(f"The total rent for Suite room(s) is: {total_rent}")
else:
    print("Invalid room type")
```
