```python
# Base class
class Vehicle:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def display_info(self):
        print(f"{self.year} {self.brand} {self.model}")

    def vehicle_type(self):
        return "General Vehicle"


# Subclass for Car
class Car(Vehicle):
    def __init__(self, brand, model, year, num_doors, fuel_type):
        # Initialize parent class (Vehicle)
        super().__init__(brand, model, year)
        self.num_doors = num_doors
        self.fuel_type = fuel_type

    def vehicle_type(self):
        return "Car"

    def car_features(self):
        return f"Number of doors: {self.num_doors}, Fuel type: {self.fuel_type}"


# Subclass for Bike
class Bike(Vehicle):
    def __init__(self, brand, model, year, has_gears):
        # Initialize parent class (Vehicle)
        super().__init__(brand, model, year)
        self.has_gears = has_gears

    def vehicle_type(self):
        return "Bike"

    def bike_features(self):
        return f"Has gears: {self.has_gears}"


# Creating objects of Car and Bike
car = Car("Toyota", "Corolla", 2020, 4, "Petrol")
bike = Bike("Yamaha", "FZ", 2021, True)

# Displaying information about Car and Bike
print("Car Information:")
car.display_info()
print(f"Type: {car.vehicle_type()}")
print(car.car_features())
print()

print("Bike Information:")
bike.display_info()
print(f"Type: {bike.vehicle_type()}")
print(bike.bike_features())

```

```python
#___WRITE PYTHON PROGRAM TO CREATE A ADVANCED CALCULATER USING SINGLE INHERITANCE___:-

import math

# Base class: Simple Calculator
class SimpleCalculator:
    def add(self, x, y):
        return x + y

    def subtract(self, x, y):
        return x - y

    def multiply(self, x, y):
        return x * y

    def divide(self, x, y):
        if y != 0:
            return x / y
        else:
            return "Error! Division by zero."

# Derived class: Advanced Calculator
class AdvancedCalculator(SimpleCalculator):
    def exponent(self, x, y):
        return x ** y

    def square_root(self, x):
        if x >= 0:
            return math.sqrt(x)
        else:
            return "Error! Square root of negative number."

    def logarithm(self, x, base=10):
        if x > 0:
            return math.log(x, base)
        else:
            return "Error! Logarithm of non-positive number."

    def sine(self, angle_degrees):
        angle_radians = math.radians(angle_degrees)
        return math.sin(angle_radians)

    def cosine(self, angle_degrees):
        angle_radians = math.radians(angle_degrees)
        return math.cos(angle_radians)

    def tangent(self, angle_degrees):
        angle_radians = math.radians(angle_degrees)
        return math.tan(angle_radians)

# Main function to use the calculator
def main():
    calc = AdvancedCalculator()

    print("Advanced Calculator")
    print("Select an operation:")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Exponent")
    print("6. Square Root")
    print("7. Logarithm")
    print("8. Sine")
    print("9. Cosine")
    print("10. Tangent")

    choice = input("Enter the number corresponding to the operation: ")

    if choice == '1':
        x = float(input("Enter first number: "))
        y = float(input("Enter second number: "))
        print(f"Result: {calc.add(x, y)}")

    elif choice == '2':
        x = float(input("Enter first number: "))
        y = float(input("Enter second number: "))
        print(f"Result: {calc.subtract(x, y)}")

    elif choice == '3':
        x = float(input("Enter first number: "))
        y = float(input("Enter second number: "))
        print(f"Result: {calc.multiply(x, y)}")

    elif choice == '4':
        x = float(input("Enter first number: "))
        y = float(input("Enter second number: "))
        print(f"Result: {calc.divide(x, y)}")

    elif choice == '5':
        x = float(input("Enter base: "))
        y = float(input("Enter exponent: "))
        print(f"Result: {calc.exponent(x, y)}")

    elif choice == '6':
        x = float(input("Enter number: "))
        print(f"Result: {calc.square_root(x)}")

    elif choice == '7':
        x = float(input("Enter number: "))
        base = float(input("Enter base (default is 10): ") or 10)
        print(f"Result: {calc.logarithm(x, base)}")

    elif choice == '8':
        angle = float(input("Enter angle in degrees: "))
        print(f"Result: {calc.sine(angle)}")

    elif choice == '9':
        angle = float(input("Enter angle in degrees: "))
        print(f"Result: {calc.cosine(angle)}")

    elif choice == '10':
        angle = float(input("Enter angle in degrees: "))
        print(f"Result: {calc.tangent(angle)}")

    else:
        print("Invalid input. Please choose a valid operation.")

if __name__ == "__main__":
    main()

```
