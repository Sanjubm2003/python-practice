```python
def even_or_odd(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"
number = 5
result = even_or_odd(number)
print(f"The number {number} is {result}.")
```

```python
names_ages = {'om': 40,
              'kartik': 25,
              'rahul': 45,
              'manju': 30
             }
age_to_names = {age: [name for name, a in names_ages.items() if a == age] for age in set(names_ages.values())}
print(age_to_names)
```

```python
def find_max(userArray):
    return max(userArray)

print("Enter the three elements followedby spaces")
userArray = map(int , input().split())
max_num = find_max(userArray)
print("The maximum number is:", max_num)
```
